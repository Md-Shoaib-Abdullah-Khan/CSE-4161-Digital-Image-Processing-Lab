```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Import filters

```python
sobelX = [[-1, 0, 1],
         [-2, 0, 2],
         [-1, 0, 1]]

sobelY = [[-1, -2, -1],
         [0, 0, 0],
         [1, 2, 1]]

prewitX = [[-1,0,1],
           [-1,0,1],
           [-1,0,1]]

prewitY = [[-1,-1,-1],
           [0, 0, 0],
           [1, 1, 1]]

laplacian = [[0,-1,0],
             [-1,4,-1],
             [0,-1,0]]
```

### Apply Laplacian filter

```python
def apply_laplacian(image, kernel):
    filtered_image = image.copy()
    offset = kernel.shape[0]//2
    height, width = image.shape
    for i in range(height):
        for j in range(width):
            pixel = 0
            for x in range(-offset, offset+1):
                for y in range(-offset, offset+1):
                    if (i+x>=0 and i+x<height and j+y>=0 and j+y<width):
                        pixel += image[i+x][j+y] * kernel[x+offset][y+offset]
            
            filtered_image[i][j] = pixel

    return filtered_image
```

### Apply Sobel filter

```python
def apply_sobel(image, kernel):
    filtered_image = image.copy()
    offset = kernel.shape[0]//2
    height, width = image.shape
    for i in range(height):
        for j in range(width):
            pixel = 0
            for x in range(-offset, offset+1):
                for y in range(-offset, offset+1):
                    if (i+x>=0 and i+x<height and j+y>=0 and j+y<width):
                        pixel += image[i+x][j+y] * kernel[x+offset][y+offset]
            
            filtered_image[i][j] = pixel

    return filtered_image
```

### Apply Prewit filter

```python
def apply_prewit(image, kernel):
    filtered_image = image.copy()
    offset = kernel.shape[0]//2
    height, width = image.shape
    for i in range(height):
        for j in range(width):
            pixel = 0
            for x in range(-offset, offset+1):
                for y in range(-offset, offset+1):
                    if (i+x>=0 and i+x<height and j+y>=0 and j+y<width):
                        pixel += image[i+x][j+y] * kernel[x+offset][y+offset]
            
            filtered_image[i][j] = pixel

    return filtered_image
```

### Load image

```python
original_image = cv2.imread("./Lincoln 221x269.tif", 0)
plt.imshow(original_image, cmap = "gray")
plt.title("Original Image")
plt.show()
```

### Laplacian filtering

```python
laplacian_filtered_image = apply_laplacian(original_image, np.array(laplacian))

plt.imshow(laplacian_filtered_image, cmap = "gray")
plt.title("Laplacian Filtered Image")
plt.show()
```

### Sobel filtering

```python
sobel_filtered_image = np.sqrt(apply_sobel(original_image, np.array(sobelX))**2 + apply_sobel(original_image, np.array(sobelY))**2)/np.sqrt(2)

plt.imshow(sobel_filtered_image, cmap = "gray")
plt.title("Sobel Filtered Image")
plt.show()
```

### Prewit filtering

```python
prewit_filtered_image = np.sqrt(apply_prewit(original_image, np.array(prewitX))**2 + apply_prewit(original_image, np.array(prewitY))**2)/np.sqrt(2)

plt.imshow(prewit_filtered_image, cmap = "gray")
plt.title("Prewit Filtered Image")
plt.show()
```
