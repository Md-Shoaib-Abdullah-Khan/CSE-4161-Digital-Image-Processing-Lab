```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Erosion function

```python
def erosion(image, structuring_element):
    eroded_image = image.copy()
    structuring_element = structuring_element*255
    offset = structuring_element.shape[0]//2
    height, width = image.shape
    for i in range(height):
        for j in range(width):
            fit = True
            for x in range(-offset, offset):
                for y in range(-offset, offset):
                    if (i+x>=0 and i+x<height and j+y>=0 and j+y<width):
                        if structuring_element[x+offset][y+offset]>0 and structuring_element[x+offset][y+offset] != image[i+x][j+y]:
                            fit = False
                    elif structuring_element[x+offset][y+offset]>0:
                        fit = False
            if fit:
                eroded_image[i][j] = 255
            else:
                eroded_image[i][j] = 0
    return eroded_image
```

### Dilation function

```python
def dilation(image, structuring_element):
    dilated_image = image.copy()
    structuring_element = structuring_element*255
    offset = structuring_element.shape[0]//2
    height, width = image.shape
    for i in range(height):
        for j in range(width):
            hit = False
            for x in range(-offset, offset):
                for y in range(-offset, offset):
                    if (i+x>=0 and i+x<height and j+y>=0 and j+y<width):
                        if structuring_element[x+offset][y+offset]>0 and structuring_element[x+offset][y+offset] == image[i+x][j+y]:
                            hit = True

            if hit:
                dilated_image[i][j] = 255
            else:
                dilated_image[i][j] = 0
                
    return dilated_image
```

### Opening function

```python
def opening(image, structuring_element):
    
    return dilation(erosion(image, structuring_element), structuring_element)
```

### Dilation function

```python
def closing(image, structuring_element):
    
    return erosion(dilation(image, structuring_element), structuring_element)
```

### Creating structuring element

```python
size = 3
structuring_element = np.ones((size, size))
```

### Loading original image

```python
noisy_fingerprint = cv2.imread("Noisy Fingerprint 315x238.tif", 0)
plt.imshow(noisy_fingerprint, cmap = "gray")
plt.title("Noisy Fingerprint Image")
plt.show()
```

### Applying Opening

```python
opened_image = opening(noisy_fingerprint, structuring_element)
plt.imshow(opened_image, cmap = "gray")
plt.title("Image after Open operation")
plt.show()
```

### Applying Closing

```python
closed_image = closing(opened_image, structuring_element)
plt.imshow(closed_image, cmap = "gray")
plt.title("Noisy Fingerprint Image")
plt.show()
```
