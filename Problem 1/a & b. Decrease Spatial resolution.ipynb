```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
originalImg = cv2.imread("lena.jpg", cv2.IMREAD_GRAYSCALE)
```

```python
def decreaseResolution(img):
    height, width = img.shape
    decreasedImg = np.zeros((height//2, width//2))
    
    for i in range(0, height, 2):
        for j in range(0, width, 2):
            decreasedImg[i//2][j//2] = img[i][j]
    
    return decreasedImg
```

```python
duplicatedImg = originalImg.copy()
```

```python
plt.figure(figsize=(15,15))

for i in range(1,5):
    plt.subplot(2,2,i)
    plt.imshow(duplicatedImg, cmap='gray')
    height, width = duplicatedImg.shape
    plt.title(f'{height}x{width}')
    duplicatedImg = decreaseResolution(duplicatedImg)

plt.show()
```
