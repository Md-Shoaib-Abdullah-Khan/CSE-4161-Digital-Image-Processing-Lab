```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
originalImg = cv2.imread("skull.jpg", cv2.IMREAD_GRAYSCALE)
```

```python
originalImg.shape
```

```python
originalImg
```

```python
def decreaseIntensityResolution(img, bits):
    height, width = img.shape
    decreasedImg = img.copy()
    steps = 255/(2**bits-1)
    
    for i in range(height):
        for j in range(width):
            decreasedImg[i][j] = round(decreasedImg[i][j]/steps) * steps
    
    return decreasedImg
```

```python
duplicatedImg = originalImg.copy()
```

```python
plt.figure(figsize=(13,8))

for i in range(8,0,-1):
    plt.subplot(2,4,9-i)
    duplicatedImg = decreaseIntensityResolution(duplicatedImg,i)
    plt.imshow(duplicatedImg, cmap='gray')
    plt.title(f'{i} bits')
    

plt.show()
```
