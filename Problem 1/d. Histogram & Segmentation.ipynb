```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
originalImg = cv2.imread('lena.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
originalImg
```

```python
def histogram(img):
    count = np.zeros(256)
    height, width = img.shape

    for i in range(height):
        for j in range(width):
            count[img[i][j]]+=1
    return count
```

```python
def segmentation(img, threshold):
    duplicateImg = img.copy()
    duplicateImg = np.where(duplicateImg < threshold, 0, 255)
    return duplicateImg
```

```python
plt.figure(figsize=(15,15))


plt.subplot(2,2,1)
plt.imshow(originalImg, cmap='gray')
plt.title('Original image')

plt.subplot(2,2,2)
plt.bar(range(256), histogram(originalImg), width=1, color='blue')
plt.title('Original image histogram')

threshold = 70
segmentedImg = segmentation(originalImg, threshold)
plt.subplot(2,2,3)
plt.imshow(segmentedImg, cmap='gray')
plt.title(f'Segmented image at Threshold {threshold}')

plt.subplot(2,2,4)
plt.bar(range(256), histogram(segmentedImg), width=1, color='blue')
plt.title(f'Segmented image histogram at Threshold {threshold}')
```
