```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
originalImg = cv2.imread('Fractured Spine 746x976.tif', cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(10,5))
plt.imshow(originalImg, cmap='gray')
plt.show()
```

```python
editedImg = originalImg.copy()

start, end, encreasedBrightness = 10, 50, 40

height, width = originalImg.shape

for i in range(height):
    for j in range(width):
        if editedImg[i][j] >= start and editedImg[i][j] <=end :
            editedImg[i][j] = editedImg[i][j] + encreasedBrightness
```

```python
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original image")
plt.imshow(originalImg, cmap='gray')
plt.subplot(1,2,2)

plt.title("Edited image")
plt.imshow(editedImg, cmap='gray')
plt.show()
```
