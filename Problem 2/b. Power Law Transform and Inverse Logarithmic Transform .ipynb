```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt
```

# Function of Power Law Transformation

```python
def power_law_transform(image, gamma):
    height, width = image.shape
    transformedImg = image.copy()
    for i in range(height):
        for j in range(width):
            pixel = float(image[i][j]) / 255.0
            transformedImg[i][j] = (pixel ** gamma) * 255

    return transformedImg
```

# Function of Inverse Log Transformation

```python
def inverse_log_transform(image):

    c = 255 / np.log(256)
    transformedImg = np.exp(image/c)-1

    return transformedImg
```

```python
originalImg = cv2.imread('Fractured Spine 746x976.tif', cv2.IMREAD_GRAYSCALE)

gamma_values = [1, 0.6, 0.4, 0.3]

plt.figure(figsize=(14,17))
for index, gamma in enumerate(gamma_values):
    plt.subplot(2, 2, index+1)
    plt.title(f'Power Law Transformation with Y = {gamma}')
    transformedImg = power_law_transform(originalImg, gamma)
    plt.imshow(transformedImg, cmap='gray')
plt.show()
```

```python
originalImg = cv2.imread('Aerial Image 765x769.tif', cv2.IMREAD_GRAYSCALE)

gamma_values = [1, 3, 4, 5]

plt.figure(figsize=(14,17))
for index, gamma in enumerate(gamma_values):
    plt.subplot(2, 2, index+1)
    plt.title(f'Power Law Transformation with Y = {gamma}')
    transformedImg = power_law_transform(originalImg, gamma)
    plt.imshow(transformedImg, cmap='gray')
plt.show()
```

```python
originalImg = cv2.imread('Aerial Image 765x769.tif', cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(5,10))
plt.title('Inverse Log Transformation')
transformedImg = inverse_log_transform(originalImg)
plt.imshow(transformedImg, cmap='gray')
plt.show()
```
