```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
originalImg = cv2.imread('Dollar 1192x500.tif', cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(5,10))
plt.title('Original image')
plt.imshow(originalImg, cmap = 'gray')
plt.show()
```

```python
mask = '11100000'
transformedImg = originalImg & int(mask, 2)

plt.figure(figsize=(5,10))
plt.title('Image with last 3 msb bits values')
plt.imshow(transformedImg, cmap = 'gray')
plt.show()
```

```python
plt.figure(figsize=(5,10))
plt.title('Difference of both images')
plt.imshow(originalImg - transformedImg, cmap = 'gray')
plt.show()
```
